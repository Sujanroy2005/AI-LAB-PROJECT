1. Develop a python program to implement a Rule-Based expert system for Student Performance Evaluation. The system should take inputs such as attendence, marks, assignment performence, and and internal assignmebnt and use IF THEN rules to classify the student as Exelent, Good, Average, or Poor. 

```python
print("STUDENT PERFORMANCE EVALUATION SYSTEM")

attendance = float(input("Enter attendance percentage: "))
marks = float(input("Enter marks percentage: "))
assignment = float(input("Enter assignment performance percentage: "))
internal = float(input("Enter internal assignment percentage: "))

if attendance >= 90 and marks >= 85 and assignment >= 85 and internal >= 85:
    print("Student Performance: Excellent")

elif attendance >= 80 and marks >= 70 and assignment >= 70 and internal >= 70:
    print("Student Performance: Good")

elif attendance >= 70 and marks >= 50 and assignment >= 50 and internal >= 50:
    print("Student Performance: Average")

else:
    print("Student Performance: Poor")
```

2. Devlop a python program to implement a rule-based expert system for medical diagnosis. The system should accept symptoms such as fever, cough, headache, and score throat and use IF THEN rules to identify a possible disease.

```python
print("MEDICAL DIAGNOSIS EXPERT SYSTEM")

fever = input("Do you have fever? (yes/no): ")
cough = input("Do you have cough? (yes/no): ")
headach = input("Do you have headache? (yes/no): ")
sore_throat = input("Do you have sore throat? (yes/no): ")

# Rule 1
if fever == "yes" and cough == "yes" and headach == "yes" and sore_throat == "yes":
    print("Possible disease: Flu")

# Rule 2
elif cough == "yes" and sore_throat == "yes" and fever == "no":
    print("Possible disease: Common Cold")

# Rule 3
elif fever == "yes" and sore_throat == "yes" and cough == "no":
    print("Possible disease: Strep Throat")

# Rule 4
elif fever == "yes" and headach == "yes":
    print("Possible disease: Viral Infection")

# Rule 5
else:
    print("Disease could not be identified.")
```
