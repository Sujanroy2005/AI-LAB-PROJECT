Write a Python program to implement the A* search algorithm to find the shortest/optimal path from a start node to a goal node in a weighted graph using heuristic values.

```python
import heapq


# Node class
class Node:
    def __init__(self, name, parent, g, h):
        self.name = name
        self.parent = parent
        self.g = g
        self.h = h
        self.f = g + h

    def __lt__(self, other):
        return self.f < other.f


# A* Algorithm
def astar(graph, heuristics, start, goal):

    open_list = []
    closed_set = set()

    # Create start node
    start_node = Node(start, None, 0, heuristics[start])

    # Add start node to open list
    heapq.heappush(open_list, start_node)

    while open_list:

        # Get node with lowest f value
        current_node = heapq.heappop(open_list)

        # Check if goal is reached
        if current_node.name == goal:
            path = []

            while current_node:
                path.append(current_node.name)
                current_node = current_node.parent

            return path[::-1]

        # Add current node to closed set
        closed_set.add(current_node.name)

        # Check all neighboring nodes
        for neighbor, cost in graph[current_node.name]:

            if neighbor in closed_set:
                continue

            # Calculate g value
            g = current_node.g + cost

            # Get heuristic value
            h = heuristics.get(neighbor, 0)

            # Create neighbor node
            neighbor_node = Node(neighbor, current_node, g, h)

            # Add neighbor to open list
            heapq.heappush(open_list, neighbor_node)

    return None


# Example graph and heuristic
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 5), ('E', 12)],
    'C': [('F', 3)],
    'D': [],
    'E': [('G', 3)],
    'F': [('E', 2)],
    'G': []
}


heuristics = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 1,
    'E': 5,
    'F': 2,
    'G': 0
}


# Start and goal
start = 'A'
goal = 'G'


# Run A* algorithm
path = astar(graph, heuristics, start, goal)


# Display result
if path:
    print("A* Path:", " -> ".join(path))
else:
    print("No path found.")
```
