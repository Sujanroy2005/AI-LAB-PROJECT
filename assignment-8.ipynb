```python
import math

def minimax(depth, node_index, is_maximize, values, max_depth):
    # Base case: target depth reached
    if depth == max_depth:
        return values[node_index]
        
    if is_maximize:
        return max(
            minimax(depth + 1, node_index * 2, False, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, False, values, max_depth)
        )
    else:
        return min(
            minimax(depth + 1, node_index * 2, True, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, True, values, max_depth)
        )

# Driver code (Moved outside the function)
leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]
tree_depth = int(math.log2(len(leaf_values)))
best_score = minimax(0, 0, True, leaf_values, tree_depth)
print("Best possible score:", best_score)
```

2.write a python program to implement the alpha beta pruning algorithm for a two-player game. use the leaf node values 3,5,2,9,12,5,23,23. determine the optimal score for tha max player and display which branch are pruned during the search.

```python
import math

def alpha_beta(depth, node_index, maximize_player, values, alpha, beta, max_depth):
    # 1. Base Case: Target depth reached (Leaf node)
    if depth == max_depth:
        return values[node_index]

    if maximize_player:
        best = float('-inf')
        # Traverse the 2 child nodes
        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                False,
                values,
                alpha,
                beta,
                max_depth
            )
            best = max(best, value)
            alpha = max(alpha, best)
            
            # Alpha-Beta Pruning condition
            if beta <= alpha:
                print(f"Branch pruned at depth {depth}, node index {node_index}")
                break
        return best

    else:
        best = float('inf')
        # Traverse the 2 child nodes
        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                True,
                values,
                alpha,
                beta,
                max_depth
            )
            best = min(best, value)
            beta = min(beta, best)
            
            # Alpha-Beta Pruning condition
            if beta <= alpha:
                print(f"Branch pruned at depth {depth}, node index {node_index}")
                break
        return best

# --- Driver Code to Test the Algorithm ---
if __name__ == "__main__":
    # Sample leaf values (8 elements = depth 3 binary tree)
    leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]
    tree_depth = int(math.log2(len(leaf_values)))
    
    # Initialize Alpha as -infinity and Beta as +infinity
    initial_alpha = float('-inf')
    initial_beta = float('inf')
    
    # Run the algorithm starting from the root (node 0)
    best_score = alpha_beta(0, 0, True, leaf_values, initial_alpha, initial_beta, tree_depth)
    print("\nBest possible score:", best_score)
```
